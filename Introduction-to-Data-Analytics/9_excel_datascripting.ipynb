```python
import pandas as pd
import numpy as np
import openpyxl
# Assume 'df' is the dataframe we've been working on in Labs 1-7
# Let's create the specialized sheets for the Excel Labs

# 1. Student Data (Lab 9)
students = pd.DataFrame({
    'Student ID': range(101, 111),
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva', 'Frank', 'Grace', 'Henry', 'Ivy', 'Jack'],
    'Age': np.random.randint(15, 26, 10),
    'Gender': np.random.choice(['Male', 'Female'], 10),
    'Math': np.random.randint(40, 100, 10),
    'Science': np.random.randint(40, 100, 10),
    'English': np.random.randint(40, 100, 10),
    'Computer': np.random.randint(40, 100, 10)
})

# 2. Sales Data (Lab 10 & 11)
regions = ['North', 'South', 'East', 'West']
products = ['Laptop', 'Tablet', 'Smartphone', 'Monitor']
sales_data = pd.DataFrame({
    'Date': pd.date_range(start='2023-01-01', periods=100, freq='D'),
    'Region': [np.random.choice(regions) for _ in range(100)],
    'Product': [np.random.choice(products) for _ in range(100)],
    'Sales Amount': np.random.uniform(500, 5000, 100),
    'Quantity': np.random.randint(1, 10, 100),
    'Revenue': 0 # To be calculated in Excel
})

# Export to a single Excel file with multiple sheets
with pd.ExcelWriter('Lab_Final_Submission.xlsx') as writer:
    students.to_excel(writer, sheet_name='Student_System', index=False)
    sales_data.to_excel(writer, sheet_name='Sales_Analysis', index=False)
    # Adding a messy sheet for Lab 9 Text Processing
    pd.DataFrame({'Raw_Name': [' john DOE', 'jane_smith ', '  ALICE w.'], 
                  'Phone': ['1234567890', '9876543210', '5551234444']}).to_excel(writer, sheet_name='Messy_Data', index=False)

print("Master Excel file 'Lab_Final_Submission.xlsx' is ready for Labs 9-11.")
```
