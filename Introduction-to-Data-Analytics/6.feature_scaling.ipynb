```python

from sklearn.preprocessing import MinMaxScaler
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("retail_data.csv")
scalar = MinMaxScaler()
```

```python
df['Annual_Income'] = df['Annual_Income'].fillna(0)
df['Income_data_fit'] = scalar.fit_transform(df[['Annual_Income']])

plt.figure(figsize=(12,5))
plt.subplot(1,2,1)
sns.histplot(df["Income_data_fit"], kde=True, color='green')
plt.title("Minmax scaling")

plt.figure(figsize=(12,5))
plt.subplot(1,2,1)
sns.histplot(df["Annual_Income"], kde=True, color='green')
plt.title("Original Data")
```

```python
mean = df["Spending_Score"].mean()
max = df['Spending_Score'].max()
min = df['Spending_Score'].min()

df['normalized_spendings'] = (df['Spending_Score'] - mean)/(max- min)
```

```python
sns.histplot(x = df['normalized_spendings'], kde=True, color= 'blue')
```

```python
from sklearn.preprocessing import MaxAbsScaler

scalar = MaxAbsScaler()
df['max-abs-spending'] = scalar.fit_transform(df[['Spending_Score']])

sns.histplot(df['max-abs-spending'], kde=True, color='purple')
print(df)
plt.show()
```
