```python
class Solution:
    def isMatch(self, s:str, p:str) -> bool:
        memo = {}
        
        def dp(i, j):
            if (i,j) in memo: return memo[(i,j)]

            if j == len(p):
                return i == len(s)

            first_match = i< len(s) and p[j] in {s[i], '.'}

            if j + 1 < len(p) and p[j+1] == '*':
                res = dp(i, j+2) or (first_match and dp(i+1, j))

            else:
                res = first_match and dp(i+1, j+1)
            
            memo[(i,j)] = res

            return res
        return dp(0,0)
```

```python
class Solution:
    def maxArea(self, height: List[int]) -> int:
        l, r =0, len(height) - 1
        max_area = 0

        while l<r:
            curr_h = min(height[l] , height[r])
            curr_w = r -l
            max_area = max(max_area, curr_h* curr_w)

            if height[l] < height[r]:
                l+=1
            else:
                r-=1
        return max_area
```

```python
import collections
from typing import List
class Solution:
    def orangesRotting(self, grid: List[List[int]]) -> int:
        ROWS, COLS = len(grid), len(grid[0])

        fresh, time = 0,0
        q = collections.deque()

        for r in range(ROWS):
            for c in range(COLS):
                if grid[r][c] == 1: fresh +=1
                if grid[r][c] == 2: q.append((r,c))
        
        directions = [[0,1], [ 0, -1], [1,0], [-1,0]]

        while q and fresh > 0:
            for i in range(len(q)):
                r, c = q.popleft()
                for dr, dc in directions:
                    row, col = r+dr, c+dc
                    if(row in range(ROWS) and col in range(COLS) and grid[row][col] ==1):
                        grid[row][col] = 2
                        q.append((row, col))
                        fresh -=1
            time +=1
        return time if fresh == 0 else -1
```

```python
class Solution:
    def threeSum(self, nums):
        nums.sort()
        result = []

        for i in range(len(nums) -2):
            if i>0 and nums[i] == nums[i-1]:
                continue

            left, right = i+1, len(nums) -1
            while left < right:
                current_sum = nums[i] +nums[left] + nums[right]
                if current_sum < 0:
                    left +=1
                elif current_sum >0:
                    right -=1
                else:
                    result.append([nums[i], nums[left], nums[right]])

                    while left < right and nums[left] == nums[left +1]:
                        left +=1
                    
                    while left < right and nums[right] == nums[right -1]:
                        right -=1

                    left +=1
                    right -=1
        
        return result
```

```python
class Solution:
    def letterCombinations(self, digits: str) -> List[str]:
        if not digits:
            return []
        
        # Mapping of digits to letters
        phone_map = {
            '2': 'abc',
            '3': 'def', 
            '4': 'ghi',
            '5': 'jkl',
            '6': 'mno',
            '7': 'pqrs',
            '8': 'tuv',
            '9': 'wxyz'
        }
        
        result = []
        
        def backtrack(index, current):
            # If we've processed all digits, add the current combination
            if index == len(digits):
                result.append(current)
                return
            
            # Get letters for current digit
            digit = digits[index]
            letters = phone_map[digit]
            
            # Try each possible letter for this digit
            for letter in letters:
                backtrack(index + 1, current + letter)
        
        # Start backtracking from index 0 with empty string
        backtrack(0, "")
        
        return result
```

```python
class Solution:
    def removeNthFromEnd(self, head: Optional[ListNode], n: int) -> Optional[ListNode]:
        ahead = head
        behind = head
        
        # Move ahead n steps forward
        for _ in range(n):
            ahead = ahead.next
        
        # If ahead is None, we need to remove head
        if not ahead:
            return head.next
        
        # Move both until ahead reaches last node
        while ahead.next:
            ahead = ahead.next
            behind = behind.next
        
        # Remove nth node from end
        behind.next = behind.next.next
        
        return head
```

```python
class Solution:
    def removeNthFromEnd(self, head:Optional[listNode], n: int) -> Optional[ListNode]:
        if not head.next:
            return None
        lead = head
        follow = head

        for _ in range(n):
            lead = lead.next
        
        if not lead:
            return head.next

        while lead.next:
            lead = lead.next
            follow = follow.next

        follow.next = follow.next.next

        return head
```

```python
class Solution:
    def isValid(self, s: str) -> bool:
        stack = []
        pairs = {'(': ')', '[': ']', '{': '}'}
        
        for char in s:
            if char in pairs:  # Opening bracket
                stack.append(char)
            else:  # Closing bracket
                if not stack or pairs[stack.pop()] != char:
                    return False
        
        return not stack  # True if stack is empty
```

```python
class Solution:
    def isValid(self, s: str) -> bool:
        stack = []
        pairs = {'(': ')', '[': ']', '{': '}'}

        for char in s:
            if char in pairs:
                stack.append(char)
            else:
                if not stack or pairs[stack.pop()] != char:
                    return False

        return not stack
```
