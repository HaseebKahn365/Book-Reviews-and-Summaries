```python
class DetectSquares:

    def __init__(self):
        self.points_count = {}
        self.x_to_y = {}  

    def add(self, point):
        x, y = point
        key = (x, y)
        self.points_count[key] = self.points_count.get(key, 0) + 1
        
        if x not in self.x_to_y:
            self.x_to_y[x] = []
        self.x_to_y[x].append(y)

    def count(self, point):
        qx, qy = point
        total = 0
        

        if qx not in self.x_to_y:
            return 0
        
        for y1 in self.x_to_y[qx]:
            if y1 == qy:
                continue  
            d = abs(y1 - qy)
            
        
            for dx in [d, -d]:
                x1 = qx + dx

                count1 = self.points_count.get((x1, qy), 0)
                count2 = self.points_count.get((x1, y1), 0)
                count3 = self.points_count.get((qx, y1), 0)
                total += count1 * count2 * count3
        
        return total
```

```python
class Solution:
    def numDistinct(self, s: str, t: str) -> int:
        memo = {}
        
        def dp(i, j):
            # Base cases
            if j == len(t):  # All characters of t matched
                return 1
            if i == len(s):  # Reached end of s but not all t matched
                return 0
            
            # Check memo
            if (i, j) in memo:
                return memo[(i, j)]
            
            # If characters match, we have two choices:
            # 1. Use this match (move both pointers)
            # 2. Skip this character in s (look for other matches)
            if s[i] == t[j]:
                result = dp(i + 1, j + 1) + dp(i + 1, j)
            else:
                # Characters don't match, only skip s character
                result = dp(i + 1, j)
            
            memo[(i, j)] = result
            return result
        
        return dp(0, 0)
```

```python
from collections import deque
from typing import List

class Solution:
    def ladderLength(self, beginWord: str, endWord: str, wordList: List[str]) -> int:
        # Convert wordList to set for O(1) lookups
        word_set = set(wordList)
        
        # If endWord not in wordList, no transformation possible
        if endWord not in word_set:
            return 0
        
        # BFS queue: (word, distance)
        queue = deque([(beginWord, 1)])
        
        while queue:
            word, distance = queue.popleft()
            
            # If we reached endWord
            if word == endWord:
                return distance
            
            # Try changing each character position
            for i in range(len(word)):
                # Try all possible letters
                for char in 'abcdefghijklmnopqrstuvwxyz':
                    if char == word[i]:
                        continue
                    
                    # Generate new word
                    new_word = word[:i] + char + word[i+1:]
                    
                    # If new_word exists in word_set
                    if new_word in word_set:
                        # Remove from set to avoid revisiting
                        word_set.remove(new_word)
                        queue.append((new_word, distance + 1))
        
        return 0
```

```python
class Solution:
    def canCompleteCircuit(self, gas: List[int], cost: List[int]) -> int:
        total_gas = 0
        current_gas = 0
        start_index = 0
        
        for i in range(len(gas)):
            # Calculate net gain/loss for this station
            net = gas[i] - cost[i]
            total_gas += net
            current_gas += net
            
            # If we run out of gas, reset starting point
            if current_gas < 0:
                start_index = i + 1
                current_gas = 0
        
        # If total gas >= total cost, solution exists
        return start_index if total_gas >= 0 else -1
```

```python
class TrieNode:
    def __init__(self):
        self.children = {}
        self.word = None  # Store the word at end node

class Solution:
    def findWords(self, board: List[List[str]], words: List[str]) -> List[str]:
        # Build Trie
        root = TrieNode()
        for word in words:
            node = root
            for char in word:
                if char not in node.children:
                    node.children[char] = TrieNode()
                node = node.children[char]
            node.word = word  # Store word at end
        
        m, n = len(board), len(board[0])
        result = []
        
        def dfs(i, j, node):
            char = board[i][j]
            
            # Check if char exists in Trie
            if char not in node.children:
                return
            
            # Move to next node
            next_node = node.children[char]
            
            # If we found a word
            if next_node.word:
                result.append(next_node.word)
                next_node.word = None  # Avoid duplicates
            
            # Mark as visited
            board[i][j] = '#'
            
            # Explore neighbors
            for di, dj in [(1,0), (-1,0), (0,1), (0,-1)]:
                ni, nj = i + di, j + dj
                if 0 <= ni < m and 0 <= nj < n and board[ni][nj] != '#':
                    dfs(ni, nj, next_node)
            
            # Backtrack
            board[i][j] = char
            
            # Optimization: remove leaf node if empty
            if not next_node.children:
                del node.children[char]
        
        # Search from each cell
        for i in range(m):
            for j in range(n):
                dfs(i, j, root)
        
        return result
```

```python
class Solution:
    def invertTree(self, root: Optional[TreeNode]) -> Optional[TreeNode]:
        if not root:
            return None
        
        stack = [root]
        
        while stack:
            node = stack.pop()
            
            # Swap children
            node.left, node.right = node.right, node.left
            
            # Add children to stack
            if node.left:
                stack.append(node.left)
            if node.right:
                stack.append(node.right)
        
        return root
```

```python
class Solution:
    def longestCommonSubsequence(self, text1: str, text2: str) -> int:
        m, n = len(text1), len(text2)
        
        # DP table: dp[i][j] = LCS of text1[0:i] and text2[0:j]
        dp = [[0] * (n + 1) for _ in range(m + 1)]
        
        for i in range(1, m + 1):
            for j in range(1, n + 1):
                if text1[i - 1] == text2[j - 1]:
                    # Characters match
                    dp[i][j] = dp[i - 1][j - 1] + 1
                else:
                    # Take max of skipping char from text1 or text2
                    dp[i][j] = max(dp[i - 1][j], dp[i][j - 1])
        
        return dp[m][n]
```

```python
import heapq

class Solution:
    def networkDelayTime(self, times: List[List[int]], n: int, k: int) -> int:
        # Build adjacency list
        graph = [[] for _ in range(n + 1)]
        for u, v, w in times:
            graph[u].append((v, w))
        
        # Dijkstra's
        distances = [float('inf')] * (n + 1)
        distances[k] = 0
        
        heap = [(0, k)]
        
        while heap:
            current_dist, node = heapq.heappop(heap)
            
            if current_dist > distances[node]:
                continue
            
            for neighbor, weight in graph[node]:
                distance = current_dist + weight
                if distance < distances[neighbor]:
                    distances[neighbor] = distance
                    heapq.heappush(heap, (distance, neighbor))
        
        # Check if all nodes reachable
        max_time = max(distances[1:])
        
        return max_time if max_time < float('inf') else -1
```

```python
class Solution:
    def minCostClimbingStairs(self, cost: List[int]) -> int:
        n = len(cost)
        
        # Track min cost for previous two steps
        prev2 = 0  # dp[i-2]
        prev1 = 0  # dp[i-1]
        
        for i in range(2, n + 1):
            # Calculate current min cost
            current = min(prev1 + cost[i-1], prev2 + cost[i-2])
            
            # Update for next iteration
            prev2, prev1 = prev1, current
        
        return prev1
```

```python
class Solution:
    def productExceptSelf(self, nums: List[int]) -> List[int]:
        n = len(nums)
        
        # Prefix products
        prefix = [1] * n
        for i in range(1, n):
            prefix[i] = prefix[i-1] * nums[i-1]
        
        # Suffix products  
        suffix = [1] * n
        for i in range(n-2, -1, -1):
            suffix[i] = suffix[i+1] * nums[i+1]
        
        # Combine
        answer = [prefix[i] * suffix[i] for i in range(n)]
        return answer
```

```python
from collections import defaultdict, deque

class Solution:
    def findCheapestPrice(self, n: int, flights: List[List[int]], src: int, dst: int, k: int) -> int:
        # Build graph: from_city -> [(to_city, price)]
        graph = defaultdict(list)
        for from_city, to_city, price in flights:
            graph[from_city].append((to_city, price))
        
        # BFS with cost tracking and stops limit
        # Queue: (current_city, total_cost, stops_made)
        queue = deque([(src, 0, 0)])
        
        # Track minimum cost to reach each city with given stops
        min_cost = [float('inf')] * n
        min_cost[src] = 0
        
        while queue:
            city, cost, stops = queue.popleft()
            
            # If we've made too many stops, skip
            if stops > k:
                continue
            
            # Explore neighbors
            for neighbor, price in graph[city]:
                new_cost = cost + price
                # Only consider if cheaper path found
                if new_cost < min_cost[neighbor]:
                    min_cost[neighbor] = new_cost
                    queue.append((neighbor, new_cost, stops + 1))
        
        return min_cost[dst] if min_cost[dst] != float('inf') else -1
```

```python
class Solution:
    def maxProfit(self, prices: List[int]) -> int:
        if not prices:
            return 0
        
        # Track three states
        hold = -prices[0]  # Max profit with stock
        sold = 0           # Max profit just sold
        rest = 0           # Max profit in cooldown/no stock
        
        for i in range(1, len(prices)):
            prev_hold = hold
            prev_sold = sold
            
            # Update states
            hold = max(hold, rest - prices[i])   # Keep holding or buy
            sold = prev_hold + prices[i]         # Sell today
            rest = max(rest, prev_sold)          # Cooldown
        
        return max(sold, rest)
```

```python
class Solution:
    def maxCoins(self, nums: List[int]) -> int:
        # Add boundary 1's
        nums = [1] + nums + [1]
        n = len(nums)
        
        # dp[i][j] = max coins from bursting balloons i+1 to j-1
        # (i and j are boundaries, not included in bursting)
        dp = [[0] * n for _ in range(n)]
        
        # Length of interval (number of balloons to burst)
        for length in range(2, n):  # At least 2 boundaries
            for left in range(n - length):
                right = left + length
                
                # Try each balloon as last to burst in this interval
                for k in range(left + 1, right):
                    # Burst balloon k last in interval (left, right)
                    coins = nums[left] * nums[k] * nums[right]
                    coins += dp[left][k] + dp[k][right]
                    dp[left][right] = max(dp[left][right], coins)
        
        return dp[0][n-1]
```

```python
from collections import defaultdict

class Solution:
    def findItinerary(self, tickets: List[List[str]]) -> List[str]:
        # Build adjacency list, sort destinations for lexical order
        graph = defaultdict(list)
        for from_city, to_city in tickets:
            graph[from_city].append(to_city)
        
        # Sort destinations in reverse so we pop from end (O(1))
        for city in graph:
            graph[city].sort(reverse=True)
        
        itinerary = []
        
        def dfs(city):
            # Visit all outgoing flights
            while graph[city]:
                next_city = graph[city].pop()
                dfs(next_city)
            
            # Add to itinerary in reverse order
            itinerary.append(city)
        
        # Start from JFK
        dfs("JFK")
        
        # Reverse to get correct order
        return itinerary[::-1]
```

```python
import heapq
from collections import Counter

def topKFrequent_heap(nums, k):
    # Count frequencies
    freq = Counter(nums)
    
    # Use min-heap of size k
    heap = []
    
    for num, count in freq.items():
        heapq.heappush(heap, (count, num))
        if len(heap) > k:
            heapq.heappop(heap)
    
    # Return the elements (ignore counts)
    return [num for count, num in heap]
```

```python
from collections import Counter

def isNStraightHand_sorting(hand, groupSize):
    if len(hand) % groupSize != 0:
        return False
    
    # Count frequency of each card
    count = Counter(hand)
    
    # Sort the unique cards
    sorted_cards = sorted(count.keys())
    
    for card in sorted_cards:
        if count[card] > 0:
            # Try to form a group starting from this card
            freq = count[card]
            for i in range(groupSize):
                current_card = card + i
                if count[current_card] < freq:
                    return False
                count[current_card] -= freq
    
    return True
```

```python
def getSum(a, b):
    """
    Add two integers without using + or -
    Using bit manipulation
    Time: O(1) - maximum 32 iterations
    Space: O(1)
    """
    # 32-bit mask to handle negative numbers
    mask = 0xFFFFFFFF
    max_int = 0x7FFFFFFF
    
    while b != 0:
        # Calculate carry (bits where both a and b have 1)
        carry = (a & b) << 1
        
        # Calculate sum without carry (XOR operation)
        a = (a ^ b) & mask
        
        # Update b with carry for next iteration
        b = carry & mask
    
    # For Python, handle negative numbers
    return a if a <= max_int else ~(a ^ mask)
```

```python
def carFleet_simple(target, position, speed):
    """
    Simplified version without explicit stack
    Time: O(n log n), Space: O(n)
    """
    n = len(position)
    if n == 0:
        return 0
    
    # Create and sort cars by position (descending)
    cars = sorted(zip(position, speed), reverse=True)
    
    # Track the time of the current leading fleet
    fleets = 0
    current_time = 0
    
    for pos, spd in cars:
        time_needed = (target - pos) / spd
        
        # If this car takes longer than the car in front,
        # it forms a new fleet
        if time_needed > current_time:
            fleets += 1
            current_time = time_needed
    
    return fleets
```

```python
import math

def minEatingSpeed(piles, h):
    """
    Binary search for minimum eating speed
    Time: O(n log m) where m is max pile size
    Space: O(1)
    """
    # Minimum possible speed is 1
    left = 1
    # Maximum speed needed is the largest pile
    right = max(piles)
    
    # Helper function to check if a speed is valid
    def can_finish(speed):
        total_hours = 0
        for pile in piles:
            # Ceiling division: math.ceil(pile / speed)
            total_hours += math.ceil(pile / speed)
        return total_hours <= h
    
    # Binary search for minimum valid speed
    while left < right:
        mid = (left + right) // 2
        
        if can_finish(mid):
            # Current speed works, try smaller speed
            right = mid
        else:
            # Current speed too slow, try faster
            left = mid + 1
    
    return left
```

```python
class Solution:
    def pacificAtlantic(self, heights: List[List[int]]) -> List[List[int]]:
        if not heights:
            return []
        
        m, n = len(heights), len(heights[0])

        pacific = [[False] * n for _ in range(m)]
        atlantic = [[False] * n for _ in range(m)]
        
        directions = [(1,0), (-1,0), (0,1), (0,-1)]
        
        def dfs(i, j, visited, prev_height):

            if i < 0 or i >= m or j < 0 or j >= n or visited[i][j] or heights[i][j] < prev_height:
                return
            
            visited[i][j] = True
            
           
            for di, dj in directions:
                dfs(i + di, j + dj, visited, heights[i][j])
        

        for i in range(m):
            dfs(i, 0, pacific, heights[i][0])
        for j in range(n):
            dfs(0, j, pacific, heights[0][j]) 

        for i in range(m):
            dfs(i, n - 1, atlantic, heights[i][n - 1])  
        for j in range(n):
            dfs(m - 1, j, atlantic, heights[m - 1][j])

        result = []
        for i in range(m):
            for j in range(n):
                if pacific[i][j] and atlantic[i][j]:
                    result.append([i, j])
        
        return result
```

```python
import heapq

class Solution:
    def minInterval(self, intervals: List[List[int]], queries: List[int]) -> List[int]:

        intervals.sort()
        

        sorted_queries = sorted([(q, i) for i, q in enumerate(queries)])
        

        min_heap = []
        
        result = [-1] * len(queries)
        interval_idx = 0
        
        for query, query_idx in sorted_queries:

            while interval_idx < len(intervals) and intervals[interval_idx][0] <= query:
                start, end = intervals[interval_idx]
                size = end - start + 1
                heapq.heappush(min_heap, (size, end))
                interval_idx += 1
            
            while min_heap and min_heap[0][1] < query:
                heapq.heappop(min_heap)

            if min_heap:
                result[query_idx] = min_heap[0][0]
        
        return result
```

```python
class Solution:
    def findTargetSumWays(self, nums: List[int], target: int) -> int:
        total = sum(nums)
        
        # If target is out of possible range
        if abs(target) > total:
            return 0
        
        # Let P = sum of numbers with '+', N = sum with '-'
        # We want: P - N = target
        # And: P + N = total
        # Adding: 2P = target + total
        # So: P = (target + total) / 2
        # We need to find subsets with sum = P
        
        if (target + total) % 2 != 0:
            return 0
        
        subset_sum = (target + total) // 2
        
        # DP: dp[i] = ways to achieve sum i
        dp = [0] * (subset_sum + 1)
        dp[0] = 1  # Empty subset
        
        for num in nums:
            # Iterate backwards to avoid reuse
            for i in range(subset_sum, num - 1, -1):
                dp[i] += dp[i - num]
        
        return dp[subset_sum]
```
