```python
def addTwoNumbers( l1, l2):
    print("start")
    curr=0
    res = []

    while(len(l1) > curr and len(l2) < curr):
        print(f"current in both{curr}")
        curr+=1
        res.append(l1[curr] + l2[curr])
    
    while(len(l1) <= curr):
        print(f"current only in l1: ", curr)
        res.append(l1[curr])
        curr+=1
    
    while(len(l2) <= curr):
        print(f"current only in l2: {curr}")
        res.append(l2[curr])
        curr+=1
    return res
    
```

```python
class Solution:
    def lengthOfLongestSubstring(self, s: str) -> int:
        d = {}
        count=0
        max = 0
        for c in s:
            if d[c] is not None:
                count+=1
                d[c] = " "
                if max<count : 
                    max = count
            else:
                count =0
                d = {}

            
```

```python
sol = Solution()
# Test strings and expected outputs
tests = [
    ("", 0),
    ("a", 1),
    ("abcdef", 6),
    ("abcabcbb", 3),   # "abc"
    ("bbbbb", 1),      # "b"
    ("pwwkew", 3),     # "wke"
    ("dvdf", 3),       # "vdf"
    ("abba", 2),       # "ab" or "ba"
]
# Run and print results
for s, expected in tests:
    result = sol.lengthOfLongestSubstring(s)
    print(f'Input: "{s}" | Expected: {expected} | Got: {result}')
```
